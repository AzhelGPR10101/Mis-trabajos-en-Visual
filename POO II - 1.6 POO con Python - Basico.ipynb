## 1. La idea: el molde y las galletas

Una **clase** es un **molde**: describe cómo será algo, pero no es ese algo.
Un **objeto** es lo que sale del molde: algo concreto, con sus propios datos.

| Molde (clase) | Lo que sale del molde (objetos) |
|---|---|
| Molde de galletas | galleta 1, galleta 2, galleta 3 |
| Plano de una casa | la casa de Ana, la casa de Luis |
| `Libro` | «Python para todos», «Aprende Java» |

Todo objeto tiene:
- **Atributos:** lo que el objeto **tiene** (sus datos). Un libro tiene título, autor y estado.
- **Métodos:** lo que el objeto **hace** (sus acciones). Un libro se puede prestar y devolver.

> En el Sistema de Biblioteca guardábamos cada libro en un **diccionario**. Ahora cada libro será un **objeto**: los datos y las acciones quedan juntos.

## 2. La clase más simple

Se define con `class` y el nombre empieza con **mayúscula**. Para crear un objeto se escribe el nombre de la clase con paréntesis, como si fuera una función.

```python
class Libro:
    pass          # pass = "todavía no tiene nada"

libro1 = Libro()  # crear un objeto (en Java: new Libro())
libro2 = Libro()

print(type(libro1))
print(libro1 is libro2)   # son dos objetos distintos
```

## 3. El constructor `__init__` y los atributos

`__init__` se ejecuta **automáticamente** cada vez que se crea un objeto. Sirve para darle sus datos iniciales.

- `self` es **el propio objeto** que se está creando (en Java se llama `this`). Siempre es el primer parámetro.
- `self.titulo = titulo` guarda el dato **dentro** del objeto.

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo          # atributo
        self.autor = autor            # atributo
        self.estado = "Disponible"    # atributo con valor inicial fijo

libro1 = Libro("Python para todos", "Charles Severance")
libro2 = Libro("Aprende Java", "Ana Torres")

print(libro1.titulo, "-", libro1.autor, "-", libro1.estado)
print(libro2.titulo, "-", libro2.autor, "-", libro2.estado)
```

Fíjese en que al crear el objeto **no se escribe `self`**: Python lo pasa solo.
`Libro("Python para todos", "Charles Severance")` llena `titulo` y `autor`.

Los atributos se leen y se cambian con un **punto**: `objeto.atributo`.

```python
libro2.autor = "Ana María Torres"    # cambiar un atributo
print(libro2.autor)
print(libro1.autor)                  # libro1 no cambia: cada objeto tiene sus propios datos
```

## 4. Métodos: lo que el objeto hace

Un método es una **función escrita dentro de la clase**. También recibe `self` como primer parámetro, y con él puede leer y cambiar los atributos del objeto.

```python
class Libro:
    def __init__(self, titulo, autor):
        self.titulo = titulo
        self.autor = autor
        self.estado = "Disponible"

    def mostrar(self):
        print(f"{self.titulo} ({self.autor}) - {self.estado}")

    def prestar(self):
        if self.estado == "Disponible":
            self.estado = "Prestado"
            print(f"Se prestó: {self.titulo}")
        else:
            print(f"No se puede: {self.titulo} ya está prestado")

    def devolver(self):
        self.estado = "Disponible"
        print(f"Se devolvió: {self.titulo}")


libro1 = Libro("Python para todos", "Charles Severance")
libro1.mostrar()
libro1.prestar()
libro1.prestar()      # segunda vez: ya está prestado
libro1.mostrar()
libro1.devolver()
libro1.mostrar()
```

Para usar un método también se usa el punto: `libro1.prestar()`.
Compare con la versión de diccionarios: antes se escribía `prestar_libro(libros, prestamos)`; ahora **se le pide al objeto** que se preste.

## 5. Métodos que devuelven un valor

Un método puede devolver un resultado con `return`, igual que una función.

```python
class Rectangulo:
    def __init__(self, base, altura):
        self.base = base
        self.altura = altura

    def area(self):
        return self.base * self.altura

    def perimetro(self):
        return 2 * (self.base + self.altura)


r = Rectangulo(5, 3)
print("Área:", r.area())
print("Perímetro:", r.perimetro())
```

## 6. Varios objetos en una lista

Como cada objeto guarda sus propios datos, se pueden poner muchos en una lista y recorrerlos con `for`.

```python
biblioteca = [
    Libro("Python para todos", "Charles Severance"),
    Libro("Aprende Java", "Ana Torres"),
    Libro("Bases de datos", "Carlos Mora"),
]

biblioteca[1].prestar()

print("\nLibros de la biblioteca:")
for libro in biblioteca:
    libro.mostrar()

disponibles = 0
for libro in biblioteca:
    if libro.estado == "Disponible":
        disponibles += 1
print("Disponibles:", disponibles)
```

## 7. Mostrar un objeto con `print`: el método `__str__`

Si se imprime un objeto directamente, Python muestra algo poco claro, como `<__main__.Libro object at 0x...>`.
El método especial `__str__` define el texto que se ve con `print` (en Java es `toString()`).

```python
class Estudiante:
    def __init__(self, nombre, nota):
        self.nombre = nombre
        self.nota = nota

    def aprobo(self):
        return self.nota >= 7

    def __str__(self):
        return f"{self.nombre}: {self.nota}"


e = Estudiante("Ana", 9)
print(e)                 # usa __str__
print("¿Aprobó?", e.aprobo())
```

## Resumen

| Concepto | En Python | En Java |
|---|---|---|
| Clase (el molde) | `class Libro:` | `class Libro { }` |
| Constructor | `def __init__(self, titulo):` | `public Libro(String titulo)` |
| El propio objeto | `self` | `this` |
| Atributo | `self.titulo = titulo` | `this.titulo = titulo;` |
| Crear un objeto | `Libro("Python")` | `new Libro("Python")` |
| Usar un método | `libro.prestar()` | `libro.prestar();` |
| Texto del objeto | `__str__` | `toString()` |

**Regla para recordar:** la clase dice **qué tiene** (atributos) y **qué hace** (métodos); cada objeto guarda **sus propios datos**.
